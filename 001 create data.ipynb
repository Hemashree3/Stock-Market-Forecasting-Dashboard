```python
import yfinance as yf
import pandas as pd
import numpy as np
```

```python
tickers = ["MSFT", "TSLA", "GOOG", "META"]
start = "2022-08-19"
end   = "2025-12-19"

df = yf.download(tickers, start=start, end=end, interval="1d", auto_adjust=False, group_by="ticker")
```

```python
df
```

```python
df.columns
```

```python
tesla = df[('TSLA', 'Close')].values
msft = df[('MSFT',     'Close')].values
google = df[('GOOG',     'Close')].values
meta = df[('META',     'Close')].values
tesla
```

```python
ind = df.index
ind
```

```python
formats = [msft, google, meta, tesla]
formats
```

```python
data = pd.DataFrame(formats)
data = data.transpose()
data.columns = ['msft', 'google', 'meta', 'tesla']
data
```

```python
data = data.set_index(ind)
data
```

```python
data.to_csv("stock_data.csv")
```
